### Step 3: Handle Missing Data


- Identify missing values in the dataset.

- Impute or remove missing values as necessary to ensure data completeness.



```python
pd.set_option('display.max_rows', None)
missing_values = df.isnull().sum()
print(missing_values)
```

```python
import numpy as np

def convert_years(val):
    if pd.isnull(val):
        return np.nan
    val = str(val).strip()
    if val.lower() == "less that 1 year":
        return 0.5
    elif val.lower() == "more that 50 years":
        return 51
    else:
        try:
            return float(val)
        except:
            return np.nan

df['YearsCodePro'] = df['YearsCodePro'].apply(convert_years)
df['YearsCodePro'].fillna(df['YearsCodePro'].median(), inplace=True)

remote_work_mode = df['RemoteWork'].mode()[0]
df['RemoteWork'].fillna(remote_work_mode, inplace=True)

job_satisfaction_mean = df['JobSat'].mean()
df['JobSat'].fillna(job_satisfaction_mean, inplace=True)

df[['LanguageHaveWorkedWith', 'LanguageWantToWorkWith']].dropna(inplace=True)
```

### Step 4: Analyze Key Columns


- Examine key columns such as `Employment`, `JobSat` (Job Satisfaction), and `YearsCodePro` (Professional Coding Experience).

- **Instruction**: Calculate the value counts for each column to understand the distribution of responses.



```python
# Employment distribution
print("Employment Distribution:\n")
print(df['Employment'].value_counts(dropna=False))

# Job Satisfaction distribution
print("\nJobSat Distribution:\n")
print(df['JobSat'].value_counts(dropna=False))

# Years of Professional Coding Experience distribution
print("\nYearsCodePro Distribution:\n")
print(df['YearsCodePro'].value_counts(dropna=False))
```

### Step 5: Visualize Job Satisfaction (Focus on JobSat)


- Create a pie chart or KDE plot to visualize the distribution of `JobSat`.

- Provide an interpretation of the plot, highlighting key trends in job satisfaction.


```python
job_sat_counts = df['JobSat'].value_counts()

plt.figure(figsize=(10, 10))
plt.pie(job_sat_counts, labels=job_sat_counts.index, autopct='%1.1f%%', startangle=90)
plt.title("Job Satisfaction Distribution (Pie Chart)")
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.kdeplot(df['JobSat'].dropna(), fill=True, color='blue')
plt.title("Job Satisfaction Distribution (KDE Plot)")
plt.xlabel("JobSat Score")
plt.ylabel("Density")
plt.show()
```

### Step 6: Programming Languages Analysis


- Compare the frequency of programming languages in `LanguageHaveWorkedWith` and `LanguageWantToWorkWith`.
  
- Visualize the overlap or differences using a Venn diagram or a grouped bar chart.


```python
# Split and explode LanguageHaveWorkedWith
df_have = df.dropna(subset=['LanguageHaveWorkedWith']).copy()
df_have = df_have.assign(Language=df_have['LanguageHaveWorkedWith'].str.split(';')).explode('Language')
df_have['Language'] = df_have['Language'].str.strip()

# Split and explode LanguageWantToWorkWith
df_want = df.dropna(subset=['LanguageWantToWorkWith']).copy()
df_want = df_want.assign(Language=df_want['LanguageWantToWorkWith'].str.split(';')).explode('Language')
df_want['Language'] = df_want['Language'].str.strip()
```

```python
have_counts = df_have['Language'].value_counts()
want_counts = df_want['Language'].value_counts()

# combine for comparison
lang_compare= pd.DataFrame({
    'HaveWorkedWith': have_counts,
    'WantToWorkWith': want_counts
})
```

```python
top_languages = lang_compare.head(10)
top_languages.plot(kind='bar', figsize=(12,6))
plt.title("Top Programming Languages: Worked With vs Want To Work With")
plt.xlabel("Programming Language")
plt.ylabel("Number of Respondents")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

### Step 7: Analyze Remote Work Trends


- Visualize the distribution of RemoteWork by region using a grouped bar chart or heatmap.


```python
#Count RemoteWork response by region
remote_region_ct = pd.crosstab(df['Country'], df['RemoteWork'], normalize='index') * 100
remote_region_ct.head()
```

```python
selected_regions = ['South Africa', 'Algeria', 'Angola', 'Zimbabwe']
subset = remote_region_ct.loc[selected_regions]

subset.plot(kind='bar', figsize=(12,6))
plt.title("Work Preferences by Region")
plt.xlabel("Region")
plt.ylabel("Percentage (%)")
plt.xticks(rotation=0)
plt.legend(title="Work Status")
plt.show()
```

### Step 8: Correlation between Job Satisfaction and Experience


- Analyze the correlation between overall job satisfaction (`JobSat`) and `YearsCodePro`.
  
- Calculate the Pearson or Spearman correlation coefficient.


```python
# Pearson Corrolation
pearson_corr = df['YearsCodePro'].corr(df['JobSat'], method='pearson')
print(f"Pearson corrolation: {pearson_corr}")

# Spearman Correlation
spearman_corr = df['YearsCodePro'].corr(df['JobSat'], method='spearman')
print(f"Spearmancorrelation: {spearman_corr}")
```

### Step 9: Cross-tabulation Analysis (Employment vs. Education Level)


- Analyze the relationship between employment status (`Employment`) and education level (`EdLevel`).

- **Instruction**: Create a cross-tabulation using `pd.crosstab()` and visualize it with a stacked bar plot if possible.


```python
# Cross-tabulate EdLevel vs Employment
edu_emp_ct = pd.crosstab(df['EdLevel'], df['Employment'], normalize='index') * 100

plt.figure(figsize=(12, 6))
edu_emp_ct.plot(kind='bar', stacked=True)
plt.title("Employment Type Distribution by Education Level")
plt.xlabel("Education Level")
plt.ylabel("Percentage (%)")
plt.xticks(rotation=45)
plt.legend(title="Employment Type", bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```
