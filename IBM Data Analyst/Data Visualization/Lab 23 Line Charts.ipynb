### Task 1: Trends in Compensation Over Age Groups


##### 1. Line Chart of Median `ConvertedCompYearly` by Age Group


- Track how the median yearly compensation (ConvertedCompYearly) changes across different age groups.

- Use a line chart to visualize these trends.


```python
# Keep relevant columns
age_comp = df[['Age', 'ConvertedCompYearly']].dropna()

# Calculate median compensation by age group
median_comp= (
    age_comp.groupby('Age')['ConvertedCompYearly'].median()
)

# Order age group correctly
age_order = [
    'Under 18 years old',
    '18-24 years old',
    '25-34 years old',
    '35-44 years old',
    '45-54 years old',
    '55-64 years old',
    '65 years or older'
]

median_comp = median_comp.reindex(age_order)

plt.plot(
    median_comp.index,
    median_comp.values,
    marker='o'
)

plt.title('Median Yearly Compensation by Age Group')
plt.xlabel('Age Group')
plt.ylabel('Median ConvertedCompYearly')
plt.xticks(rotation=45)
for x, y in zip(median_comp.index, median_comp.values):
    plt.text(x, y, f'{y:,.0f}', ha='center', va='bottom')
plt.show()
```

##### 2. Line Chart of Median `ConvertedCompYearly` for Ages 25 to 45


For a closer look, plot a line chart focusing on the median compensation for respondents between ages 25 and 45.


```python
# Filter ages 25 - 44
age_fltd = df[
    df['Age'].isin([
        '25-34 years old',
        '35-44 years old'
    ])
][['Age', 'ConvertedCompYearly']].dropna()

median_comp = (
    age_fltd.groupby('Age')['ConvertedCompYearly'].median()
)

plt.plot(
    median_comp.index,
    median_comp.values,
    marker='x'
)
plt.title('Median Compensation for Respondents Aged 25-44')
plt.xlabel('Age Group')
plt.ylabel('Median Compensation (ConvertedCompYearly)')
for x, y in zip(median_comp.index, median_comp.values):
    plt.text(x, y, f'${y:,.0f}', ha='center', va='bottom')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

### Task 2: Trends in Job Satisfaction by Experience Level



##### 1. Line Chart of Job Satisfaction (`JobSatPoints_6`) by Experience Level



- Use a column that approximates experience level to analyze how job satisfaction changes with experience.

- If needed, substitute an available experience-related column for `Experience`.


```python
jobsat_exp = df[['WorkExp', 'JobSatPoints_6']].dropna()

median_jobsat = (
    jobsat_exp
    .groupby('WorkExp')['JobSatPoints_6']
    .median()
)

plt.figure(figsize=(15,10))

plt.plot(
    median_jobsat.index,
    median_jobsat.values,
    marker='o'
)

plt.title('Median Job Satisfaction by Experience Level')
plt.xlabel('Years of Work Experience')
plt.ylabel('Median JobSatPoints_6')
plt.grid(True, linestyle='--', alpha=0.6)

plt.show()
```

### Task 3: Trends in Job Satisfaction and Compensation by Experience


##### 1.Line Chart of Median ConvertedCompYearly Over Experience Level

- This line chart will track how median compensation (`ConvertedCompYearly`) changes with increasing experience.

- Use a column such as `WorkExp` or another relevant experience-related column.


```python
# Keep relevant columns
comp_exp = df[['WorkExp', 'ConvertedCompYearly']].dropna()

# Calculate median compensation for each experience level
median_comp = (
    comp_exp.groupby('WorkExp')['ConvertedCompYearly'].median()
)

plt.plot(
    median_comp.index,
    median_comp.values,
    marker='o'
)
plt.title('Median Compensation by Work Experience')
plt.xlabel('Years of Work Experience')
plt.ylabel('Median ConvertedCompYearly')
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```

##### 2.Line Chart of Job Satisfaction (`JobSatPoints_6`) Across Experience Levels

- Create a line chart to explore trends in job satisfaction (`JobSatPoints_6`) based on experience level.

- This chart will provide insight into how satisfaction correlates with experience over time


```python
# Keep relevant columns
jobsat_exp = df[['WorkExp', 'JobSatPoints_6']].dropna()

# Calculate median JobSatPoints_6 for each experience level
jobsat_madian = (
    jobsat_exp.groupby('WorkExp')['JobSatPoints_6'].median()
)

plt.plot(
    jobsat_madian.index,
    jobsat_madian.values,
    marker='o',
    linewidth=2,
    color='steelblue'
)

plt.title('Median Job Satisfaction by Experience Level')
plt.xlabel('Years of Work Experience')
plt.ylabel('Median JobSatPoints_6')
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```
