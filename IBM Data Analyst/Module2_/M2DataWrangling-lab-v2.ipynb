### 1. Load the Dataset


<h5>1.1 Import necessary libraries and load the dataset.</h5>


Ensure the dataset is loaded correctly by displaying the first few rows.


```python
# Import necessary libraries
import pandas as pd

# Load the Stack Overflow survey data
dataset_url = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/n01PQ9pSmiRX6520flujwQ/survey-data.csv"
df = pd.read_csv(dataset_url)

# Display the first few rows
print(df.head())
```

<h5>2.1 Summarize the dataset by displaying the column data types, counts, and missing values.</h5>


```python
print("== Dataset Info ==")
print(df.info())

print("\n== Dataset data types ==")
print(df.dtypes)

print("\n== Dataset shape ==")
print(df.shape)

print("\n== Dataset summary statistics ==")
print(df.describe(include='all'))

print("\n== Count of missing values in each column ==")
print(df.isnull().sum())
```

```python
# statistics for numerical columns
print("== Numerical Columns Summary ==")
df.describe()
```

### 3. Identifying and Removing Inconsistencies


<h5>3.1 Identify inconsistent or irrelevant entries in specific columns (e.g., Country).</h5>


```python
# Inconsistent or irrelevant data entries for "Country" column
print("== Unique values in 'Country' column ==")
print(df['Country'].unique())

country_counts = df['Country'].value_counts()
print(country_counts)
```

```python
country_counts = df['Country'].value_counts()
print(country_counts.head(30))
```

<h5>3.2 Standardize entries in columns like Country or EdLevel by mapping inconsistent values to a consistent format.</h5>


```python
# Mapping for Country
country_mapping = {
    'USA': 'United States',
    'U.S.': 'United States',
    'Untied States': 'United States',
    'UK': 'United Kingdom',
    'England': 'United Kingdom',
    'Prefer not to say': None,
    'Unknown': None
}

# Mapping for EdLevel
edlevel_mapping = {
    'Bachelors': "Bachelor’s degree",
    'BS': "Bachelor’s degree",
    'Undergraduate degree': "Bachelor’s degree",
    'Masters': "Master’s degree",
    'MS': "Master’s degree",
    'PhD': "Doctoral degree",
    'Doctorate': "Doctoral degree",
    'Some college': "Some college/university study without degree"
}

# Standardize Country
df['Country'] = df['Country'].replace(country_mapping).str.strip()

# Standardize EdLevel
df['EdLevel'] = df['EdLevel'].replace(edlevel_mapping).str.strip()

print(df['Country'].value_counts().head(10))
print(df['EdLevel'].value_counts().head(10))
```

### 4. Encoding Categorical Variables


<h5>4.1 Encode the Employment column using one-hot encoding.</h5>


```python
# Handle missing values in Employment column
most_frequent_employment = df['Employment'].mode()[0]
df['Employment'] = df['Employment'].fillna(most_frequent_employment)

# Apply one-hot encoding
employment_encoded = pd.get_dummies(df['Employment'], prefix='Employment')

# Concatenate encoded column back to the dataset
df = pd.concat([df, employment_encoded], axis=1)

# Verify the changes
df.head()
```

### 5. Handling Missing Values


<h5>5.1 Identify columns with the highest number of missing values.</h5>


```python
# Counting missing values per column
missing_values = df.isnull().sum().reset_index()
missing_values.columns = ['Column', 'MissingValues']

# Sort by highest number of missing values
missing_values = missing_values.sort_values(by='MissingValues', ascending=False)

missing_values.head(10)
```

<h5>5.2 Impute missing values in numerical columns (e.g., `ConvertedCompYearly`) with the mean or median.</h5>


```python
# Identfying numeric columns
numeric_columns = df.select_dtypes(include=['int', 'float']).columns.tolist()

missing_numeric = df[numeric_columns].isnull().sum().reset_index()
missing_numeric.columns = ["Column", "MissingValues"]

missing_numeric
```

```python
# Impute numeric values
mean_compTotal = df["CompTotal"].mean()
df['CompTotal'] = df['CompTotal'].fillna(mean_compTotal)

median_workExp = df['WorkExp'].median()
df['WorkExp'] = df['WorkExp'].fillna(median_workExp)

mean_convertedCompYearly = df['ConvertedCompYearly'].mean()
df['ConvertedCompYearly'] = df['ConvertedCompYearly'].fillna(mean_convertedCompYearly)

mean_job_satisfaction = df['JobSat'].mean()
df['JobSat'] = df['JobSat'].fillna(mean_job_satisfaction)
```

<h5>5.3 Impute missing values in categorical columns (e.g., `RemoteWork`) with the most frequent value.</h5>


```python
frequent_remote_work = df['RemoteWork'].mode()[0]
df['RemoteWork'] = df['RemoteWork'].fillna(frequent_remote_work)
```

### 6. Feature Scaling and Transformation


<h5>6.1 Apply Min-Max Scaling to normalize the `ConvertedCompYearly` column.</h5>


```python
## Write your code here
min_val = df['ConvertedCompYearly'].min()
max_val = df['ConvertedCompYearly'].max()

df['ConvertedCompYearly_MinMax'] = (df['ConvertedCompYearly'] - min_val) / (max_val - min_val)
```

<h5>6.2 Log-transform the ConvertedCompYearly column to reduce skewness.</h5>


```python
# Apply log transformation
import numpy as np
df['ConvertedCompYearly_log'] = np.log1p(df['ConvertedCompYearly'])
```

```python

import matplotlib.pyplot as plt

# Compare distributions
plt.figure(figsize=(12,5))

# Original distribution
plt.subplot(1,2,1)
df['ConvertedCompYearly'].plot(kind='hist', bins=50, color='skyblue', edgecolor='black')
plt.title("Original Compensation Distribution")
plt.xlabel("ConvertedCompYearly")

# Log-transformed distribution
plt.subplot(1,2,2)
df['ConvertedCompYearly_log'].plot(kind='hist', bins=50, color='salmon', edgecolor='black')
plt.title("Log-Transformed Compensation Distribution")
plt.xlabel("ConvertedCompYearly_Log")

plt.tight_layout()
plt.show()
```

### 7. Feature Engineering


<h5>7.1 Create a new column `ExperienceLevel` based on the `YearsCodePro` column:</h5>


```python
df['YearsCodePro'].head(50)
```

```python
# Step 1: Convert YearsCodePro to numeric safely
def convert_years(val):
    if pd.isnull(val):
        return np.nan
    val = str(val).strip()
    if val.lower() == "less than 1 year":
        return 0.5
    elif val.lower() == "more than 50 years":
        return 51
    else:
        try:
            return float(val)
        except:
            return np.nan

df['YearsCodePro'] = df['YearsCodePro'].apply(convert_years)

# Step 2: Fill missing values with median
years_code_median = df['YearsCodePro'].median()
df['YearsCodePro'].fillna(years_code_median, inplace=True)

# Step 3: Categorize into ExperienceLevel
def categorize_experience(years):
    if years is None or np.isnan(years):   # safeguard against None/NaN
        return "Unknown"
    elif years <= 1:
        return "Beginner"
    elif years <= 5:
        return "Intermediate"
    elif years <= 10:
        return "Advanced"
    else:
        return "Expert"

df['ExperienceLevel'] = df['YearsCodePro'].apply(categorize_experience)

df[['YearsCodePro', 'ExperienceLevel']].head(30)
```
